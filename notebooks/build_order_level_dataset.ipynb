```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", 100)

RAW_DATA_PATH = Path("../data/raw")
```

```python
orders = pd.read_csv(
    RAW_DATA_PATH / "olist_orders_dataset.csv",
    parse_dates=[
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
    ],
)

order_items = pd.read_csv(
    RAW_DATA_PATH / "olist_order_items_dataset.csv",
    parse_dates=["shipping_limit_date"],
)

customers = pd.read_csv(
    RAW_DATA_PATH / "olist_customers_dataset.csv"
)

products = pd.read_csv(
    RAW_DATA_PATH / "olist_products_dataset.csv"
)

sellers = pd.read_csv(
    RAW_DATA_PATH / "olist_sellers_dataset.csv"
)

payments = pd.read_csv(
    RAW_DATA_PATH / "olist_order_payments_dataset.csv"
)
```

```python
tables = {
    "orders": orders,
    "order_items": order_items,
    "customers": customers,
    "products": products,
    "sellers": sellers,
    "payments": payments,
}
```

```python
items_enriched=order_items.merge(
    products,
    on="product_id",
    how="left",
    validate="m:1",
    indicator="_product_match"

)
```

```python
items_enriched["_product_match"].value_counts()
```

```python
items_enriched.isna().sum().sort_values(ascending=False)
```

```python
items_enriched = items_enriched.drop(columns="_product_match")
```

```python
items_enriched = items_enriched.merge(
    sellers,
    on="seller_id",
    how="left",
    validate="m:1",
    indicator="_seller_match"
)
```

```python
items_enriched["_seller_match"].value_counts()
```

```python
items_enriched = items_enriched.drop(columns="_seller_match")
```

```python
print(items_enriched.shape)
print(items_enriched["order_id"].nunique())
```

```python
order_item_summary = (
    items_enriched
    .groupby("order_id", as_index=False)
    .agg(
        number_of_items=("order_item_id", "count"),
        number_of_unique_products=("product_id", "nunique"),
        number_of_unique_sellers=("seller_id", "nunique"),
        total_product_price=("price", "sum"),
        total_freight_value=("freight_value", "sum"),
        average_product_weight_g=("product_weight_g", "mean"),
        average_product_length_cm=("product_length_cm", "mean"),
        average_product_height_cm=("product_height_cm", "mean"),
        average_product_width_cm=("product_width_cm", "mean"),
        number_of_product_categories=("product_category_name", "nunique")
    )
)
```

```python
payments.nunique()
```

```python
order_level_payments = (
    payments.groupby("order_id", as_index=False)
    .agg(
        total_payment_installments=("payment_installments", "sum"),
        total_payment_value=("payment_value", "sum"),
        number_of_payment_records=("payment_sequential", "count")
    )
)
order_level_payments
```

```python
primary_payment_type = (
    payments
    .sort_values(
        ["order_id", "payment_value", "payment_sequential"],
        ascending=[True, False, True]
    )
    .drop_duplicates(subset="order_id", keep="first")
    [["order_id", "payment_type"]]
    .rename(columns={"payment_type": "primary_payment_type"})
)
```

```python
order_level_payments = order_level_payments.merge(
    primary_payment_type,
    on="order_id",
    how="left",
    validate="one_to_one"
)
order_level_payments
```
